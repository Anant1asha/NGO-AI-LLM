# 🌟 AASHA AIOS — Autonomous Cloud Worker & Add-On Infrastructure (Mode A)

Welcome to the **AASHA AIOS Autonomous Cloud Runner** on Google Colab.

### Dual-Authority Execution Architecture:
- **Mode D (Direct Local PC Workspace)**: Executes on your primary Windows workstation when active.
- **Mode A (Autonomous Cloud Worker)**: Executes here in Google Colab when your PC is offline, asleep, or idle.

### What this Cloud Worker does:
1. Clones the sovereign task branch `aios/v2-bootstrap` across `NGO-AI-LLM` and `Aasha-AI`.
2. Executes the full continuous 12-stage engineering loop (`laya:loop`).
3. Compiles offline textbook chapters with PyMuPDF vectorization.
4. Runs all 9 verification test batteries (Master E2E, L-Truth, Concurrency, Fault Hardening).
5. Pushes verified commits and built chapter single-file HTMLs strictly to `origin/aios/v2-bootstrap` (zero pushes to `master`/`main`).

---
### 🔑 Step 0: Set Up Secure Colab Secrets (Zero Secret Leakage)
Before running, add your GitHub token in the **Secrets** tab on the left sidebar (key icon):
- **Name**: `GH_PAT`
- **Value**: Your GitHub Personal Access Token (with `repo` / `contents: write` scope)
- **Notebook access**: Toggle **ON**

*(Optional)*: If you have an OpenRouter API key, add `OPENROUTER_API_KEY`.

````python
# @title 🚀 Step 1: Bootstrap Cloud Worker Environment
import os, sys, subprocess
from google.colab import userdata

print("=======================================================")
print("  🌟 AASHA AIOS: Bootstrapping Mode A Cloud Worker     ")
print("=======================================================")

# 1. Retrieve secure token from Colab Secrets
try:
    GH_PAT = userdata.get('GH_PAT')
    print("✔ GitHub PAT loaded securely from Colab Secrets.")
except Exception as e:
    raise ValueError("❌ Please add 'GH_PAT' to Colab Secrets (key icon in left sidebar) before running!")

# Optional OpenRouter key
try:
    os.environ['OPENROUTER_API_KEY'] = userdata.get('OPENROUTER_API_KEY')
    print("✔ OpenRouter API Key configured.")
except Exception:
    print("ℹ No OpenRouter key provided. Falling back to Tier 0 Offline Synthesizer & Vectorizer.")

# 2. Install Node.js 20 LTS & system utilities
print("\nInstalling Node.js 20 LTS and PyMuPDF...")
os.system("curl -fsSL https://deb.nodesource.com/setup_20.x | bash - > /dev/null 2>&1")
os.system("apt-get install -y nodejs > /dev/null 2>&1")
os.system("pip install -q pymupdf pytest pillow requests")

# Check versions
node_ver = subprocess.getoutput("node -v")
npm_ver = subprocess.getoutput("npm -v")
py_ver = sys.version.split()[0]
print(f"✔ Node.js: {node_ver} | npm: {npm_ver} | Python: {py_ver}")
````

````python
# @title 📥 Step 2: Clone / Synchronize Repositories on branch 'aios/v2-bootstrap'
WORKSPACE_DIR = "/content/NGO-AI-LLM"
BRANCH = "aios/v2-bootstrap"

REPO_URL = f"https://x-access-token:{GH_PAT}@github.com/Anant1asha/NGO-AI-LLM.git"
SUB_REPO_URL = f"https://x-access-token:{GH_PAT}@github.com/Anant1asha/Aasha-AI.git"

if not os.path.exists(WORKSPACE_DIR):
    print(f"Cloning root repository on branch '{BRANCH}'...")
    subprocess.run(["git", "clone", "--branch", BRANCH, "--single-branch", REPO_URL, WORKSPACE_DIR], check=True)
else:
    print(f"Workspace exists. Pulling latest commits on '{BRANCH}'...")
    os.chdir(WORKSPACE_DIR)
    subprocess.run(["git", "pull", "origin", BRANCH], check=True)

os.chdir(WORKSPACE_DIR)

# Configure Git identity for cloud worker
subprocess.run(["git", "config", "user.name", "AASHA AIOS Cloud Worker (Mode A)"], check=True)
subprocess.run(["git", "config", "user.email", "cloud-worker@aasha-aios.org"], check=True)

# Ensure sub-repo Aasha-AI is cloned
SUB_DIR = os.path.join(WORKSPACE_DIR, "Aasha-AI")
if not os.path.exists(SUB_DIR):
    print(f"Cloning sub-repository Aasha-AI on branch '{BRANCH}'...")
    subprocess.run(["git", "clone", "--branch", BRANCH, "--single-branch", SUB_REPO_URL, SUB_DIR], check=True)
    os.chdir(SUB_DIR)
    subprocess.run(["git", "config", "user.name", "AASHA AIOS Cloud Worker (Mode A)"], check=True)
    subprocess.run(["git", "config", "user.email", "cloud-worker@aasha-aios.org"], check=True)
    os.chdir(WORKSPACE_DIR)

print("\nInstalling npm packages in root and sub-repo...")
subprocess.run(["npm", "install"], check=True)
subprocess.run(["npm", "--prefix", "Aasha-AI", "install"], check=True)
print("✔ Repositories synchronized and dependencies ready!")
````

````python
# @title ⚙️ Step 3: Run AASHA Continuous 12-Stage Engineering Loop
import os, subprocess

os.chdir(WORKSPACE_DIR)

# Set Cloud Worker environment variables
os.environ['AASHA_EXECUTION_MODE'] = 'Mode A (Autonomous Cloud Worker - Colab)'
os.environ['NODE_ENV'] = 'test'
os.environ['MEMO0_OFFLINE'] = '1'
os.environ['ALLOW_PAID_GEMINI_FALLBACK'] = 'false'

print("\nExecuting AASHA Continuous Engineering Loop in Mode A...")
result = subprocess.run(["node", "scripts/aasha_continuous_loop.mjs", "--once"], env=os.environ)

if result.returncode == 0:
    print("\n=======================================================")
    print("  🎉 MODE A CLOUD RUN COMPLETED SUCCESSFULLY!          ")
    print("  All tests verified, chapters compiled, and committed ")
    print("  to origin/aios/v2-bootstrap.                         ")
    print("=======================================================")
else:
    print(f"❌ Execution encountered an issue (exit code: {result.returncode})")
````
