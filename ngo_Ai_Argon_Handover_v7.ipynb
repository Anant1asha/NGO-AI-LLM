### 📂 Step 1: Initialize Workspace Scanner
Let's scan the target workspace path to identify the directory structure, code files, and general layout.

```python
import os

workspace_path = '/content/drive/Othercomputers/My Laptop/NGO AI LLM'

def scan_workspace(base_path):
    if not os.path.exists(base_path):
        print(f"❌ Path does not exist: {base_path}. Please check your Google Drive mounting.")
        return

    print(f"Scanning workspace: {base_path}\n")
    for root, dirs, files in os.walk(base_path):
        # Exclude common directories to keep the output clean
        dirs[:] = [d for d in dirs if d not in ['.git', 'node_modules', '__pycache__']]
        level = root.replace(base_path, '').count(os.sep)
        indent = ' ' * 4 * level
        print(f"{indent}📁 {os.path.basename(root)}/")
        sub_indent = ' ' * 4 * (level + 1)
        for f in files[:10]:  # Limit output files per folder
            print(f"{sub_indent}📄 {f}")
        if len(files) > 10:
            print(f"{sub_indent}... and {len(files) - 10} more files.")

scan_workspace(workspace_path)
```

### 📋 Step 2.5: Map and List Workspace Files
We will examine the top-level files in `/content/drive/Othercomputers/My Laptop/NGO AI LLM` to organize development and testing tasks.

```python
import os

workspace_path = '/content/drive/Othercomputers/My Laptop/NGO AI LLM'

if not os.path.exists(workspace_path):
    print(f"❌ Path does not exist: {workspace_path}")
else:
    print(f"📂 Listing files in {workspace_path}:\n")
    items = os.listdir(workspace_path)

    # Separate files and directories
    files = []
    dirs = []
    for item in sorted(items):
        full_path = os.path.join(workspace_path, item)
        if os.path.isdir(full_path):
            if item not in ['.git', 'node_modules', '__pycache__']:
                dirs.append(item)
        else:
            files.append(item)

    print("--- Directories ---")
    for d in dirs:
        print(f"📁 {d}/")

    print("\n--- Top-level Files ---")
    for f in files:
        file_size = os.path.getsize(os.path.join(workspace_path, f))
        print(f"📄 {f} ({file_size / 1024:.2f} KB)")
```

### 🐙 Step 3: Initialize and Sync Git Repository with GH_PAT
We will now link your workspace to GitHub using your Personal Access Token (PAT) so we can fetch, pull, and track code changes securely.

```python
import os
import subprocess

# Secure fallback assignment directly inside the script block
try:
    gh_pat = userdata.get("GH_PAT")
except Exception:
    gh_pat = os.environ.get("GH_PAT", "")
os.environ['GH_PAT'] = gh_pat

workspace_path = '/content/drive/Othercomputers/My Laptop/NGO AI LLM'

if not os.path.exists(workspace_path):
    print(f"❌ Path does not exist: {workspace_path}. Make sure Google Drive is mounted and the folder path is correct.")
else:
    os.chdir(workspace_path)

    # Check if directory is already a git repository
    if not os.path.exists('.git'):
        print("Initializing new Git repository...")
        subprocess.run(["git", "init"], check=True)
    else:
        print("✔ Git repository already initialized.")

    # Configure user credentials for commits
    subprocess.run(["git", "config", "user.name", "AASHA AIOS Cloud Worker"], check=True)
    subprocess.run(["git", "config", "user.email", "cloud-worker@aasha-aios.org"], check=True)

    # Set or update the remote origin containing the token securely
    remote_url = f"https://x-access-token:{gh_pat}@github.com/Anant1asha/NGO-AI-LLM.git"

    # Check if remote already exists
    remotes = subprocess.getoutput("git remote").splitlines()
    if "origin" in remotes:
        subprocess.run(["git", "remote", "set-url", "origin", remote_url], check=True)
        print("✔ Origin remote URL updated with GH_PAT.")
    else:
        subprocess.run(["git", "remote", "add", "origin", remote_url], check=True)
        print("✔ Origin remote URL added with GH_PAT.")

    # Fetch and check status
    print("\nFetching latest status from origin...")
    res = subprocess.run(["git", "fetch", "origin"], capture_output=True, text=True)
    print("STDOUT:", res.stdout)
    print("STDERR:", res.stderr)
```

### 🚀 Step 4: Argon AIOS Handover (Closes GAP-V7-02 & GAP-V7-03 + Launches `/swarm` Loop on `aios/v2-bootstrap`)
Applies `aasha_aios_v2_bootstrap_full.bundle` from Google Drive, pushes strictly to `origin/aios/v2-bootstrap`, runs the live Fastify HTTP suite (`GAP-V7-01`), runs Playwright headless Chromium DOM verification (`GAP-V7-03`), and launches the 1,080-agent Argon `/swarm` loop.

```python
# @title ⚡ Execute Argon AIOS Handover (GAP-V7-02 Push + GAP-V7-03 Playwright + /swarm)
import os, subprocess
from google.colab import drive, userdata

drive.mount("/content/drive", force_remount=False)
BUNDLE_CANDIDATES = [
    "/content/drive/MyDrive/NGO AI LLM/aasha_aios_v2_bootstrap_full.bundle",
    "/content/drive/Othercomputers/My Laptop/NGO AI LLM/aasha_aios_v2_bootstrap_full.bundle"
]
bundle_path = next((p for p in BUNDLE_CANDIDATES if os.path.exists(p)), BUNDLE_CANDIDATES[0])
print(f"Using bundle: {bundle_path}")

repo_dir = "/content/aasha_aios_unified_repo"
if not os.path.exists(repo_dir):
    subprocess.run(["git", "clone", "https://github.com/Anant1asha/NGO-AI-LLM.git", repo_dir], check=False)

os.chdir(repo_dir)
subprocess.run(["git", "checkout", "-B", "aios/v2-bootstrap"], check=True)
subprocess.run(["git", "fetch", bundle_path, "aios/v2-bootstrap:aios/v2-bootstrap"], check=True)
subprocess.run(["git", "reset", "--hard", "aios/v2-bootstrap"], check=True)

try:
    gh_pat = userdata.get("GH_PAT")
    if gh_pat:
        subprocess.run(["git", "remote", "set-url", "origin", f"https://x-access-token:{gh_pat}@github.com/Anant1asha/NGO-AI-LLM.git"], check=True)
except Exception:
    pass

subprocess.run(["pip", "install", "-q", "playwright"], check=True)
subprocess.run(["playwright", "install", "chromium"], check=True)
subprocess.run(["chmod", "+x", "scripts/colab_argon_handover.sh"], check=True)
subprocess.run(["bash", "scripts/colab_argon_handover.sh", bundle_path], check=True)
```
